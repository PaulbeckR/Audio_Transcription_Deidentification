Prepares transcribed files into appropriate format. 

1. Transcriptions in word doc, by interviewer / interviewee, no time stamps. 
2. Split by speaker in lines, 
3. Shift first speaker (interviewer/interviewee) to SPEAKER_0 
4. Save as .txt file

Compare against audio: 

1. Add time stamps. 
2. Requires human correction. 

```python
# Obtain transcription file. 
import os
import pandas as pd 
from docx import Document

project_path = "C://Users//pisces2//Box//PISCES-PC R01 - Operations//Qualitative Data//Rachael - Transcription Project"

output_folder = project_path + "//Transcribed Data//Test//transcription_valid"




```

```python
def process_transcript(excel_file_path, output_f):

    df = Document(excel_file_path)

    base_name = os.path.splitext(os.path.basename(excel_file_path))[0]

    txt_file_path = os.path.join(output_f, f"{base_name}.txt")

    speaker_mapping = {}
    current_speaker_index = 0
    processed_lines = []

    for paragraph in df.paragraphs:
        line = paragraph.text.strip()
        if line.startswith(("Interviewer:", "Interviewee:")):
            colon_index = line.index(":")
            speaker = line[:colon_index]
            text = line[colon_index + 1:].strip()

            if speaker not in speaker_mapping:
                speaker_mapping[speaker] = f"speaker_{current_speaker_index}"
                current_speaker_index += 1
        
        

            new_speaker_label = speaker_mapping[speaker]

            processed_lines.append(f"{new_speaker_label}: {text}") 


    with open(txt_file_path, 'w', encoding='utf-8') as txt_file:
        
        txt_file.write("\n".join(processed_lines))

    print(f"Transcript processed and saved to : {txt_file_path}")


excel_file_path = "C://Users//pisces2//Box//PISCES-PC R01 - Operations//Qualitative Data//Rachael - Transcription Project//Transcribed Data//Transcriptions//PC1-1025-01_S1_2022.02.28.docx"

output_f = output_folder

process_transcript(excel_file_path, output_f)
```
